```python
df.drop(['homepage', 'tagline', 'keywords', 'production_companies'], axis=1, inplace=True)
```

```python
df['release_date'] = pd.to_datetime(df['release_date'])
```

```python
# question 1

max_voted_film = df.query('vote_count == vote_count.max()')
max_voted_film['original_title'].values[0]
```

```python
# question 2
df.dropna(subset=['cast', 'genres'], inplace=True)

l = df['cast'].str.split('|').tolist()

l_flatten = [cast for child_list in l for cast in child_list]

d = {}

for cast in l_flatten:
    if cast in d.keys():
        d[cast] += 1
    else:
        d[cast]=1

d = dict(sorted(d.items(), key=lambda x: x[1], reverse=True))
list(d.keys())[0]
```

```python
# question 2

data = df['cast'].str.cat(sep='|')
split = data.split('|')

s = pd.Series(split)
s.value_counts().head(1).index[0]
```

```python
import matplotlib.pyplot as plt

cast10 = s.value_counts().head(10)

cast10.plot(kind='bar')
# cast10
```
