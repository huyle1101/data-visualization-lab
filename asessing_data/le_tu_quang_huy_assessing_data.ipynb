# Completeness

- Các cột Browser, OS, City, Country, Referrer, bị null 100%, vậy ta sẽ drop các cột này
- Với các giá trị null của các cột How happy are you (Salary, Salary, Coworkers, Management, Upward Mobility, Learning New Things), ta sẽ điền giá trị mean, vì các cột này không có outliers (max=10, min=0 và std~=2.7)
- Với các giá trị null của cột Q12 - Highest Level of Education thì ta sẽ điền giá trị mode, vì đây là cột categorical

```python
cols = ['Browser', 'OS', 'City', 'Country', 'Referrer']

for col in cols:
    df.drop(col, axis=1, inplace=True)
```

```python
happy_cols = [
    'Q6 - How Happy are you in your Current Position with the following? (Salary)',
    'Q6 - How Happy are you in your Current Position with the following? (Work/Life Balance)',
    'Q6 - How Happy are you in your Current Position with the following? (Coworkers)',
    'Q6 - How Happy are you in your Current Position with the following? (Management)',
    'Q6 - How Happy are you in your Current Position with the following? (Upward Mobility)',
    'Q6 - How Happy are you in your Current Position with the following? (Learning New Things)',
]

for happy_col in happy_cols:
    df[happy_col] = df[happy_col].fillna(df[happy_col].mean())
```

```python
df['Q12 - Highest Level of Education'] = df['Q12 - Highest Level of Education'].fillna(df['Q12 - Highest Level of Education'].mode()[0])
```

# Validity & Accuracy
- Trong các cột có giá trị số thì cột Current Age xuất hiện 2 dòng có giá trị là 200 và 92 tuổi, ta sẽ thay thế 2 giá trị này bằng cách điền giá trị mean mới. Giá trị mean mới sẽ được tính trên toàn bộ dataset sau khi đã loại bỏ 2 giá trị này, tránh bị nó kéo lệch. Các dòng nào có tuổi dưới 70 ta vẫn giữ nguyên.

- Các cột có giá trị số khác đều đảm bảo được Validity và Accuracy, vì chúng là các cột đánh giá, đều nằm trong đoạn [0;10]

```python
new_mean = round(df[df['Q10 - Current Age']<70]['Q10 - Current Age'].mean())

df.loc[df['Q10 - Current Age'] > 70, 'Q10 - Current Age'] = new_mean
```
