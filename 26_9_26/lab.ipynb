```python
import pandas as pd
import numpy as np

patients = pd.read_csv('patients.csv')
treatments = pd.read_csv('treatments.csv')
```

```python
patients.head()

patients['given_name'] = patients['given_name'].str.lower()
patients['surname'] = patients['surname'].str.lower()
```

```python
treatments.head()
```

```python
df_merge = pd.merge(patients, treatments, how='inner',on=['surname'])
df_merge.head()
```

```python
df_merge.shape
```

```python
df_merge_out = patients.merge(treatments, how='outer')
df_merge_out.shape
```
